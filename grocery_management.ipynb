```python
# 2. Imports
from crewai import Agent, Task, Crew, LLM
```

```python
with open(r"C:\Users\Shivani Pal\Downloads\receipt.txt.txt", "r", encoding="utf-8") as f:
    receipt_text = f.read()
    
print(receipt_text)
```

```python
# 4. LLM
llm = LLM(
    model="ollama/llama3.1:8b",
    base_url="http://localhost:11434"
)
```

```python
# 5. Agent
expiration_date_search_agent = Agent(
    role="Expiration Date Estimation Specialist",
    goal="Estimate grocery expiration dates.",
    backstory="Expert in food storage and shelf life estimation.",
    llm=llm,
    verbose=True
)
```

```python
expiration_date_search_task = Task(
    description="""
Read the following grocery receipt:

{receipt_text}

Extract the grocery items.

Estimate the shelf life and expiration date for each item.

Return the result in a markdown table.
""",
    expected_output="""
A markdown table with:
- Item
- Purchase Date (if available)
- Shelf Life
- Estimated Expiration Date
""",
    agent=expiration_date_search_agent
)
```

```python
# 7. Crew
crew = Crew(
    agents=[expiration_date_search_agent],
    tasks=[expiration_date_search_task],
    verbose=True
)
```

```python
result = await crew.kickoff_async(
    inputs={
        "receipt_text": receipt_text
    }
)

print(result.raw)
```
