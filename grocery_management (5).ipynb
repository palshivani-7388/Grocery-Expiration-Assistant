```python
# 2. Imports
from crewai import Agent, Task, Crew, LLM
```

```python
# 3. Grocery Items
grocery_items = [
    {
        "name": "Milk",
        "purchase_date": "2026-08-05"
    },
    {
        "name": "Bread",
        "purchase_date": "2026-08-05"
    }
]
```

```python
# 4. LLM
llm = LLM(
    model="ollama/llama3.1:8b",
    base_url="http://localhost:11434"
)
```

```python
# 5. Agent
expiration_date_search_agent = Agent(
    role="Expiration Date Estimation Specialist",
    goal="Estimate grocery expiration dates.",
    backstory="Expert in food storage and shelf life estimation.",
    llm=llm,
    verbose=True
)
```

```python
# 6. Task
expiration_date_search_task = Task(
    description="""
Estimate the expiration date for the grocery items provided by the user.

Use the grocery_items input.
Search StillTasty (or your available knowledge/tools) for the typical shelf life.
Calculate the estimated expiration date from the purchase date.
""",
    expected_output="""
A table containing:
- Item name
- Purchase date
- Shelf life
- Estimated expiration date
""",
    agent=expiration_date_search_agent
)
```

```python
# 7. Crew
crew = Crew(
    agents=[expiration_date_search_agent],
    tasks=[expiration_date_search_task],
    verbose=True
)
```

```python
result = await crew.kickoff_async(
    inputs={
        "grocery_items": grocery_items
    }
)

print(result)
```
